## 1. Environment and checker loading

Run the next cell first. It verifies CPython 3.11, checks the bytecode magic
number before importing, and loads the checker from this notebook's folder.

```python
from pathlib import Path
import hashlib
import importlib.util
import json
import sys

EXPECTED_CACHE_TAG = "cpython-311"
EXPECTED_NUMPY = "2.4.6"
PYC_NAME = "numpy_easy_medium_answers.cpython-311.pyc"

if sys.implementation.cache_tag != EXPECTED_CACHE_TAG:
    raise RuntimeError(
        f"This checker needs {EXPECTED_CACHE_TAG}, but this kernel is "
        f"{sys.implementation.cache_tag}. Switch to Python (AIS5102)."
    )

# NumPy is imported only after the kernel check so that a wrong kernel without
# NumPy still gets the "Switch to Python (AIS5102)" message above.
import numpy as np

candidates = [Path.cwd() / PYC_NAME, Path.cwd() / "student_release" / PYC_NAME]
pyc_path = next((path for path in candidates if path.exists()), None)
if pyc_path is None:
    raise FileNotFoundError(
        f"Could not find {PYC_NAME}. Keep the notebook and .pyc together."
    )

with pyc_path.open("rb") as handle:
    pyc_magic = handle.read(4)
if pyc_magic != importlib.util.MAGIC_NUMBER:
    raise RuntimeError(
        "The .pyc magic number does not match this Python kernel. "
        "Switch to Python (AIS5102)."
    )

manifest_path = pyc_path.parent / "manifest.json"
if not manifest_path.exists():
    raise FileNotFoundError("manifest.json is required beside the trusted .pyc file.")
manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
artifact_info = manifest.get("artifacts", {}).get(PYC_NAME)
if artifact_info is None:
    raise RuntimeError(f"{PYC_NAME} is not registered in manifest.json")
pyc_hash = hashlib.sha256(pyc_path.read_bytes()).hexdigest()
if pyc_hash != artifact_info["sha256"] or pyc_path.stat().st_size != artifact_info["bytes"]:
    raise RuntimeError(
        "The compiled checker does not match manifest.json. "
        "Restore an unmodified copy from the teacher."
    )

spec = importlib.util.spec_from_file_location("numpy_easy_medium_answers", pyc_path)
if spec is None or spec.loader is None:
    raise RuntimeError(f"Could not create a loader for {pyc_path}")
checker = importlib.util.module_from_spec(spec)
spec.loader.exec_module(checker)

print("Checker loaded:", pyc_path)
print("Verified SHA-256:", pyc_hash)
print("Runtime:", checker.version_info())
if np.__version__ != EXPECTED_NUMPY:
    print(f"Warning: expected NumPy {EXPECTED_NUMPY}, found {np.__version__}")
print("Supported tasks:", checker.available_tasks()[0], "to", checker.available_tasks()[-1])
```

## 2. How input/output comparison works

The checker can give you a test input and run the teacher implementation. This
does **not** reveal the teacher's source code; it shows the expected output so
you can compare behavior.

```python
demo_task = 36
demo_args, demo_kwargs = checker.example_inputs(demo_task)
teacher_output = checker.run_reference(demo_task, *demo_args, **demo_kwargs)

print("Contract:", checker.describe(demo_task))
print("Input args:", demo_args)
print("Teacher output:", teacher_output)
```

The helper below turns a checker result into a short report. Failures are
classified as function header, return type, shape, dtype, value, structure, or
exception problems. Each report includes a next step without revealing the
teacher implementation.

```python
ISSUE_LABELS = {
    "NotCallable": "Not a function",
    "FunctionNameError": "Function name mismatch",
    "SignatureInspectionError": "Unreadable function signature",
    "SignatureError": "Function header mismatch",
    "MissingFunction": "Missing function",
    "return_type": "Return type mismatch",
    "shape": "Shape mismatch",
    "dtype": "Dtype mismatch",
    "value": "Value mismatch",
    "structure": "Return structure mismatch",
    "exception": "Your function raised an exception",
    "exception_contract": "Required exception was not raised",
}

def issue_label(item):
    key = item.get("error_subtype") or item.get("error_type", "failed")
    return ISSUE_LABELS.get(key, str(key).replace("_", " ").title())

def show_result(result):
    task_id = result["task_id"]
    if result["passed"]:
        print(f"✅ Task {task_id:03d}: all {result['total_cases']} cases passed — congratulations! 🎉")
        if task_id < 65:
            print(f"   Great work. Keep the momentum going — Task {task_id + 1:03d} is next!")
        else:
            print("   That was Task 065 — you have completed every exercise in this kit. Outstanding!")
        return result

    print(f"❌ Task {task_id:03d}: not yet correct")
    if result.get("error"):
        print("   Issue:", issue_label(result))
        print("   Details:", result["error"])
        if result.get("guidance"):
            print("   Try:", result["guidance"])
    for case in result.get("cases", []):
        if case["passed"]:
            print(f"   ✅ case {case['index']}: {case['label']}")
            continue
        print(f"   ❌ case {case['index']}: {case['label']}")
        print("       Issue:", issue_label(case))
        print("       Details:", case.get("message", "Check failed"))
        if "expected" in case:
            print("       expected:", case["expected"])
        if "actual" in case:
            print("       actual:  ", case["actual"])
        if case.get("guidance"):
            print("       Try:", case["guidance"])
    return result
```

## Troubleshooting

- **Bad magic number / wrong cache tag:** switch the kernel to **Python (AIS5102)**.
- **File not found:** keep this notebook and the `.pyc` in the same folder.
- **SignatureError:** restore the exact function header shown above the exercise.
- **Shape or dtype mismatch:** inspect `result.shape` or `result.dtype` immediately
  before the `return`, then check `axis`, `reshape`, and `astype` choices.
- **Value mismatch:** test the same logic on a small input and check operation order,
  boundaries, and whether the reduction uses the intended axis.
- **StudentException:** follow the printed **Try** suggestion, fix the function, and
  rerun only that exercise cell.
- **A cell runs forever:** use *Kernel → Interrupt Kernel*, fix the loop, and rerun.
